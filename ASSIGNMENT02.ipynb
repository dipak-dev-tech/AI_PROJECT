```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt

# load the dataset
digits = load_digits()

# database information
print("Dataset Shape:",digits.data.shape)
print("Image Shape:",digits.images.shape)
print("Target Labels: ",digits.target_names)


# Display the first image
plt.figure(figsize=(4,4))
plt.imshow(digits.images[0], cmap = "pink")
plt.title(f"Label: {digits.target[0]}")
plt.axis("off")
plt.show()




```

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt


digits = load_digits()

fig,axes = plt.subplots(2,5,figsize = (10,5))

for i,ax in enumerate(axes.flat):
    ax.imshow(digits.images[i],cmap = "Blues")
    ax.set_title(f"Digit : {digits.target[i]}")
    ax.axis("off")
    
plt.tight_layout()
plt.show()
```


***2.write a program to display 	a.number of rows and columns, 	b.columns names,	c.data types,***


```python
from sklearn.datasets import load_iris
import pandas as pd

# Load the Iris dataset
iris = load_iris(as_frame=True)

# Create a DataFrame
df = iris.frame

# Add the target column
df["target"] = iris.target

# a. Number of rows and columns
print("Number of rows and columns:")
print(df.shape)

# b. Column names
print("\nColumn Names:")
print(df.columns)

# c. Data types
print("\nData Types:")
print(df.dtypes)
```

```python
from sklearn.datasets import load_iris, load_digits

# Load the iris datasets
iris = load_iris()

# Load the digits datasets
digits = load_digits()

# Count the number of samples
print("Number of samples in Iris dataset:", len(iris.data))
print("Number of samples in Digits dataset:", len(digits.data))
```
